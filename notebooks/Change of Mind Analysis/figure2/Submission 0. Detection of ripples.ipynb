### 1. remove Ripple band artifact

```python
success = []
for animal in list_of_days_animals.keys():
    for d in list_of_days_animals[animal]:
        nwb_file_name = animal.lower() + d + '.nwb'
        nwb_copy_file_name = get_nwb_copy_filename(nwb_file_name)
        session_interval, position_interval = runSessionNames(nwb_copy_file_name)

        for s in session_interval:
            success_ = removeArtifactInFilteredData(nwb_copy_file_name, s,
                             filter_name = 'Ripple 150-250 Hz',
                             field_name = 'artifact removed',broadband_zscore = -1)
            if success_:
                success.append((nwb_copy_file_name, s))
```

### 2. Detecting ripples

```python
default_parameters = {"speed_threshold": 4, "conservative": True,
                      "minimum_duration": 0.015, "smoothing_sigma": 0.004,
                      "Kay_threshold": 2, "Karlsson_threshold": 3, "Karlsson_number": 2,
                     }
```

```python
for animal in ['julio', 'klein']:#list_of_days_animals.keys():
    for d in list_of_days_animals[animal]:
        nwb_file_name = animal.lower() + d + '.nwb'
        nwb_copy_file_name = get_nwb_copy_filename(nwb_file_name)
        session_interval, position_interval = runSessionNames(nwb_copy_file_name)
        
        for session_name in session_interval:
            e = session_name[:2]
            ripple_times, CA1TetrodeInd, CCTetrodeInd = ripple_detection_master(nwb_file_name,e,
                                                                                fieldname = 'artifact removed',
                                                                                speed_threshold = default_parameters["speed_threshold"],
                                                                                conservative = default_parameters["conservative"])
        
        
            # for the RippleTimes table
            key = {'nwb_file_name': nwb_copy_file_name, 'epoch': e,
                   'parameter_name': 'default_conservative',
                    'parameter_value': default_parameters}
            key['ripple_times'] = ripple_times.to_dict()
            RippleTimesByParameters().insert1(key,replace = True)
```
