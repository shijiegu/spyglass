### Confirmation of raw nwb edits are correct

```python
import pynwb
import numpy as np
import matplotlib.pyplot as plt
```

```python
from spyglass.common.common_behav import RawPosition
```

```python
df = (RawPosition() & {"nwb_file_name": "lewis20240105_.nwb",
                 "interval_list_name": "pos 1 valid times"}).fetch1_dataframe()
```

```python
nwb_file_name = '/stelmo/nwb/raw/lewis20240105.nwb'

with pynwb.NWBHDF5IO(nwb_file_name, 'r',load_namespaces=True) as io:
    nwbf = io.read()
    pos_raw_new = nwbf.processing['behavior']['position']
    pos_raw_new_np = np.array(pos_raw_new.spatial_series['led_0_series_1'].data)
    spat_objs = get_all_spatial_series(nwbf, incl_times=False)[0]
    
```

```python
fig, axes = plt.subplots(1,4,figsize=(24,5))
time_ind = 10000
axes[0].scatter(pos_raw_new_np[:time_ind, 0], pos_raw_new_np[:time_ind, 1])
axes[0].set_title("new spatial series")
axes[1].scatter(np.array(df.xloc1)[:time_ind], np.array(df.yloc1)[:time_ind])
axes[1].set_title("old spatial series (RawPosition)")
axes[2].scatter(pos_raw_new_np[:time_ind, 0], np.array(df.xloc1)[:time_ind])
axes[2].set_title("new 1st coord (x-axis) vs \n   old x coord (y-axis)")
axes[3].scatter(pos_raw_new_np[:time_ind, 1], np.array(df.yloc1)[:time_ind])
axes[3].set_title("new 2nd coord (x-axis) vs \n   old y coord (y-axis)")
```
